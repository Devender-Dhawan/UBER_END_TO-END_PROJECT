```python
df['Date']=pd.to_datetime(df['Date'])
```

```python
df['Month'] = df['Date'].dt.month
df['Month_Name'] = df['Date'].dt.month_name()
df['Day'] = df['Date'].dt.day
df['Day_Name'] = df['Date'].dt.day_name()
df['Quarter'] = df['Date'].dt.quarter
df['Year'] = df['Date'].dt.year
```

```python
df['Time'] = pd.to_datetime(df['Time'].astype(str))
```

```python
df['Hour'] = df['Time'].dt.hour
```

```python
df['Booking Status'].value_counts()
```

```python
completed = (df['Booking Status'] == 'Completed').sum()

completed
```

```python
completion_rate = (
    (df['Booking Status'] == 'Completed').sum()
    / len(df)
) * 100

completion_rate
```

```python
status_summary = df['Booking Status'].value_counts().reset_index()

status_summary.columns = ['Booking Status', 'Bookings']

status_summary
```

```python
plt.figure(figsize=(10,5))

sns.barplot(
    data=status_summary,
    x='Booking Status',
    y='Bookings'
)

plt.title('Uber Booking Status Distribution')
plt.xlabel('Booking Status')
plt.ylabel('Number of Bookings')
plt.xticks(rotation=20)

plt.show()
```

```python
cancelled = df['Booking Status'].isin([
    'Cancelled by Customer',
    'Cancelled by Driver'
]).sum()

cancellation_rate = cancelled / len(df) * 100

cancellation_rate
```

```python
customer_cancel_rate = (
    (df['Booking Status'] == 'Cancelled by Customer').sum()
    / len(df)
) * 100

customer_cancel_rate
```

```python
driver_cancel_rate = (
    (df['Booking Status'] == 'Cancelled by Driver').sum()
    / len(df)
) * 100

driver_cancel_rate
```

```python
Q1 = df['Booking Value'].quantile(0.25)
Q3 = df['Booking Value'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df['Booking Value'] < lower_bound) |
    (df['Booking Value'] > upper_bound)
]

len(outliers)
```

```python
plt.figure(figsize=(10,5))

plt.boxplot(
    df['Booking Value'].dropna(),
    vert=False
)

plt.title('Booking Value Outlier Analysis')
plt.xlabel('Booking Value')

plt.show()
```
